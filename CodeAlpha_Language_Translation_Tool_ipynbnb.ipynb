```python

import gradio as gr
from deep_translator import GoogleTranslator

languages = {
    "English": "en",
    "Hindi": "hi",
    "Telugu": "te",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Japanese": "ja",
    "Chinese": "zh-CN",
    "Arabic": "ar"
}

def translate_text(text, source, target):
    if not text.strip():
        return "Please enter some text."

    if source == target:
        return text

    try:
        translator = GoogleTranslator(
            source=languages[source],
            target=languages[target]
        )
        return translator.translate(text)
    except Exception as e:
        return f"Translation error: {str(e)}"

with gr.Blocks(title="AI Language Translator") as app:

    gr.Markdown("""
    # 🌍 AI Language Translation Tool
    ### Translate text instantly between multiple languages
    """)

    text_input = gr.Textbox(
        label="📝 Enter Text",
        placeholder="Type your text here...",
        lines=5
    )

    with gr.Row():

        source_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="English",
            label="Source Language"
        )

        target_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="Hindi",
            label="Target Language"
        )

    translate_button = gr.Button(
        "🔄 Translate",
        variant="primary"
    )

    output = gr.Textbox(
        label="🌐 Translated Text",
        lines=5
    )

    translate_button.click(
        fn=translate_text,
        inputs=[text_input, source_language, target_language],
        outputs=output
    )

    gr.Markdown(
        "💡 Select the languages, enter your text, and click Translate."
    )

app.launch(share=True)
```

```python

import gradio as gr
import requests

languages = {
    "English": "en",
    "Hindi": "hi",
    "Telugu": "te",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Japanese": "ja",
    "Chinese": "zh-CN",
    "Arabic": "ar"
}

def translate_text(text, source, target):
    if not text.strip():
        return "Please enter some text."

    if source == target:
        return text

    try:
        url = "https://translate.googleapis.com/translate_a/single"

        params = {
            "client": "gtx",
            "sl": languages[source],
            "tl": languages[target],
            "dt": "t",
            "q": text
        }

        response = requests.get(url, params=params, timeout=15)
        data = response.json()

        translated = "".join(
            item[0] for item in data[0] if item[0]
        )

        return translated

    except Exception as e:
        return f"Translation error: {str(e)}"


with gr.Blocks(title="AI Language Translation Tool") as app:

    gr.Markdown("""
    # 🌍 AI Language Translation Tool
    ### Translate text instantly between multiple languages
    """)

    text_input = gr.Textbox(
        label="📝 Enter Text",
        placeholder="Type your text here...",
        lines=5
    )

    with gr.Row():

        source_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="English",
            label="Source Language"
        )

        target_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="Hindi",
            label="Target Language"
        )

    translate_button = gr.Button(
        "🔄 Translate",
        variant="primary"
    )

    output = gr.Textbox(
        label="🌐 Translated Text",
        lines=5
    )

    translate_button.click(
        fn=translate_text,
        inputs=[
            text_input,
            source_language,
            target_language
        ],
        outputs=output
    )

    gr.Markdown(
        "💡 Select the languages, enter your text, and click Translate."
    )

app.launch(share=True)
```

```python

import gradio as gr
import requests

languages = {
    "English": "en",
    "Hindi": "hi",
    "Telugu": "te",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "Spanish": "es",
    "French": "fr",
    "German": "de",
    "Japanese": "ja",
    "Chinese": "zh-CN",
    "Arabic": "ar"
}

def translate_text(text, source, target):
    if not text.strip():
        return "Please enter some text."

    if source == target:
        return text

    try:
        url = "https://api.mymemory.translated.net/get"

        params = {
            "q": text,
            "langpair": f"{languages[source]}|{languages[target]}",
            "mt": "1"
        }

        response = requests.get(url, params=params, timeout=20)
        response.raise_for_status()

        data = response.json()

        translated = data["responseData"]["translatedText"]

        return translated

    except Exception as e:
        return f"Translation failed: {str(e)}"


with gr.Blocks(title="AI Language Translation Tool") as app:

    gr.Markdown("""
    # 🌍 AI Language Translation Tool
    ### Translate text instantly between multiple languages
    """)

    text_input = gr.Textbox(
        label="📝 Enter Text",
        placeholder="Type your text here...",
        lines=5
    )

    with gr.Row():

        source_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="English",
            label="Source Language"
        )

        target_language = gr.Dropdown(
            choices=list(languages.keys()),
            value="Hindi",
            label="Target Language"
        )

    translate_button = gr.Button(
        "🔄 Translate",
        variant="primary"
    )

    output = gr.Textbox(
        label="🌐 Translated Text",
        lines=5
    )

    translate_button.click(
        fn=translate_text,
        inputs=[
            text_input,
            source_language,
            target_language
        ],
        outputs=output
    )

    gr.Markdown(
        "💡 Select the languages, enter your text, and click Translate."
    )

app.launch(share=True)
```
