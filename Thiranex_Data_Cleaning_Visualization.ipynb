```python
import pandas as pd

# Load the dataset from Google Sheets
url = "https://docs.google.com/spreadsheets/d/1DAJ0NMvM0cPOpmTLHBX8RoqhnCtxhxxk-T8w2hbKqNs/export?format=csv"

df = pd.read_csv(url)

# Display the first 5 rows
df.head()
```

```python
# Check missing values in each column

print("Missing values in each column:")
print(df.isnull().sum())
```

```python

# Handle missing values using median

# Fill missing Age values with the median Age
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Salary values with the median Salary
df["Salary"] = df["Salary"].fillna(df["Salary"].median())

# Check missing values again
print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Check for duplicate rows

print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
# Remove duplicate rows

df = df.drop_duplicates()

print("Duplicate rows after cleaning:")
print(df.duplicated().sum())
```

```python
# Remove extra spaces from text columns

df["Name"] = df["Name"].str.strip()
df["Department"] = df["Department"].str.strip()

print("Extra spaces removed successfully!")
```

```python
# Convert text columns to lowercase

df["Name"] = df["Name"].str.lower()
df["Department"] = df["Department"].str.lower()

print("Text converted to lowercase!")
```

```python
# Preview the cleaned dataset

df.head()
```

```python
# Check for outliers using the IQR method

Q1 = df[["Age", "Salary"]].quantile(0.25)
Q3 = df[["Age", "Salary"]].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower bounds:")
print(lower_bound)

print("\nUpper bounds:")
print(upper_bound)
```

```python
# Find rows containing outliers

age_outliers = df[
    (df["Age"] < lower_bound["Age"]) |
    (df["Age"] > upper_bound["Age"])
]

salary_outliers = df[
    (df["Salary"] < lower_bound["Salary"]) |
    (df["Salary"] > upper_bound["Salary"])
]

print("Age outliers:")
display(age_outliers)

print("\nSalary outliers:")
display(salary_outliers)
```

```python

# Find rows containing outliers

age_outliers = df[
    (df["Age"] < lower_bound["Age"]) |
    (df["Age"] > upper_bound["Age"])
]

salary_outliers = df[
    (df["Salary"] < lower_bound["Salary"]) |
    (df["Salary"] > upper_bound["Salary"])
]

print("Age outliers:")
display(age_outliers)

print("\nSalary outliers:")
display(salary_outliers)
```

```python

# Handle salary outliers using IQR capping

df["Salary"] = df["Salary"].clip(
    lower=lower_bound["Salary"],
    upper=upper_bound["Salary"]
)

print("Salary outliers handled successfully!")

print("\nSalary range after handling outliers:")
print("Minimum:", df["Salary"].min())
print("Maximum:", df["Salary"].max())
```

```python
# Convert Join_Date to proper datetime format

df["Join_Date"] = pd.to_datetime(df["Join_Date"])

print("Join_Date converted successfully!")

# Check the data types
print("\nData types:")
print(df.dtypes)
```

```python

import matplotlib.pyplot as plt

# Count employees by department
department_counts = df["Department"].value_counts()

# Create a bar chart
plt.figure(figsize=(8, 5))
department_counts.plot(kind="bar")

plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python

# Salary distribution

plt.figure(figsize=(8, 5))

plt.hist(df["Salary"], bins=10, edgecolor="black")

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
# Age distribution

plt.figure(figsize=(8, 5))

plt.hist(df["Age"], bins=10, edgecolor="black")

plt.title("Age Distribution of Employees")
plt.xlabel("Age")
plt.ylabel("Number of Employees")

plt.tight_layout()
plt.show()
```

```python
# Calculate average salary by department

avg_salary = df.groupby("Department")["Salary"].mean()

print("Average Salary by Department:")
print(avg_salary)

# Create a bar chart
plt.figure(figsize=(8, 5))

avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# Count employees by joining date

joining_counts = df["Join_Date"].value_counts().sort_index()

# Create a line chart
plt.figure(figsize=(10, 5))

plt.plot(
    joining_counts.index,
    joining_counts.values,
    marker="o"
)

plt.title("Employees Joining Over Time")
plt.xlabel("Join Date")
plt.ylabel("Number of Employees")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select numerical columns
numeric_data = df[["Age", "Salary"]]

# Calculate correlation
correlation = numeric_data.corr()

print("Correlation Matrix:")
display(correlation)

# Create heatmap
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
